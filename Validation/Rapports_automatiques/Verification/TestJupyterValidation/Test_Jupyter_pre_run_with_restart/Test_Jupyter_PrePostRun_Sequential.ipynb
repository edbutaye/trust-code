```python
from trustutils import run

run.reset()
run.initBuildDirectory()
r=[20,21]


# arguments of these functions are path to case and name of the case (without the .data)
def pre_run(d, n):
    name=n.replace("_restart", "_pre_run")+'.data'
    size=int(n.replace("diffusion_restart_", ""))
    case = run.TRUSTCase(".", "diffusion.data").copy(name, targetDirectory=d+"/pre_run")
    case.substitute_template(dict(number=size))
    case.run()

for k in range(*r):
    run.addCaseFromTemplate("diffusion_restart.data", f"k{k}",dict(number=k, sauv=f"./pre_run/diffusion_pre_run_{k}_pb.sauv"),nbProcs=1, targetData=f"diffusion_restart_{k}.data", pre_run=pre_run)

run.printCases()
```

```python
run.runCases(verbose=True)
```
