```python
from trustutils import run

run.reset()
run.initBuildDirectory()
r=[20,40]

for k in range(*r):
    case = run.TRUSTCase(".", "diffusion.data").copy(f"diffusion_{k}_prepare.data", targetDirectory="prepare")
    case.substitute_template(dict(number=k))
    case.run()

# arguments of these functions are path to case and name of the case (without the .data)
def pre_run(d, n):
    case = run.TRUSTCase(d, n).copy(n+'_pre_run.data', targetDirectory=d+"/pre_run")
    case.run()
    case2 = run.TRUSTCase(d, n).copy(n+'_pre_run_2.data', targetDirectory=d+"/pre_run_2")
    case2.run()
def post_run(d, n):
    case = run.TRUSTCase(d, n).copy(n+'_post_run.data', targetDirectory=d+"/post_run", pre_run=pre_run)
    case.run()

# only add the first one to non-regression tests. They are all duplicates
excluNR=False
for k in range(*r):
    run.addCaseFromTemplate("diffusion.data", f"k{k}",dict(number=k),nbProcs=1, targetData=f"diffusion_{k}.data", pre_run=pre_run, post_run=post_run, excluNR=excluNR)
    excluNR=True
run.printCases()
```

```python
run.runCases(verbose=True, preventConcurrent=True)
```
